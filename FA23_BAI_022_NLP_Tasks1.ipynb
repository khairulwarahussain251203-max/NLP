## Task 07: String Validation Methods

```python
# (a)
text = "NLP is the future of AI research"
print("Starts with NLP:", text.startswith("NLP"))
print("Ends with research:", text.endswith("research"))
print("Position of future:", text.find("future"))
```

```python
# (b)
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
for t in tokens:
    if t.isalpha():
        print(t, "-> alphabetic")
    elif t.isdigit():
        print(t, "-> numeric")
    else:
        print(t, "-> mixed")
```

```python
# (c)
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words, pure_numbers, mixed_tokens = [], [], []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)
```

```python
# (d)
def validate_token(token):
    if token.isalpha():
        return "WORD"
    if token.isdigit():
        return "NUMBER"
    return "MIXED"

for t in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(t, "->", validate_token(t))
```

## Task 08: Functions and Lambda for NLP Pipelines

```python
# (a)
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

s = "  HELLO Natural Language Processing  "
print(clean(s))
print(clean_lambda(s))
```

```python
# (b)
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

print(remove_short(tokenize(clean("I love working with NLP tools in Python daily")), 3))
```

```python
# (c)
def word_lengths(tokens):
    return {t: len(t) for t in tokens}

print(word_lengths(["language", "model", "ai", "processing", "deep"]))
```

```python
# (d)
def nlp_pipeline(text, min_len):
    tokens = remove_short(tokenize(clean(text)), min_len)
    return word_lengths(tokens)

print(nlp_pipeline("  Natural Language Processing is a key part of AI  ", 3))
print(nlp_pipeline("Python makes text analysis simple and fast", 4))
```

## Task 09: map(), filter(), sorted() and Counter

```python
# (a)
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]
print(list(map(lambda w: w.lower(), words)))
print(list(map(lambda w: len(w), words)))
```

```python
# (b)
words_long = list(filter(lambda w: len(w) > 3, words))
words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print("Long words:", words_long)
print("Words with t:", words_with_t)
```

```python
# (c)
print("Alphabetical:", sorted(words))
print("Length ascending:", sorted(words, key=len))
print("Length descending:", sorted(words, key=len, reverse=True))
```

```python
# (d)
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
freq = Counter(sentence.split())
print(freq)
print(freq.most_common(3))
print('Frequency of "python":', freq["python"])
print("Total words:", sum(freq.values()))
print("Vocabulary size:", len(freq))
```
